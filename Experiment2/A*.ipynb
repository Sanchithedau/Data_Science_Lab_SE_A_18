### A* Search Algorithm

The A* search algorithm is a popular pathfinding algorithm that finds the shortest path between a starting node and a goal node in a graph. It combines features of Dijkstra's algorithm and greedy best-first search to efficiently find the optimal path.

#### Key Concepts:

1.  **Nodes**: Represent points or states in the search space.
2.  **Edges (or Transitions)**: Represent connections between nodes, each with an associated cost.
3.  **Heuristic Function `h(n)`**: An estimated cost from the current node `n` to the goal node. It should be admissible (never overestimates the actual cost).
4.  **Cost Function `g(n)`**: The actual cost from the starting node to the current node `n`.
5.  **Evaluation Function `f(n)`**: The total estimated cost of the path through node `n` to the goal. It's calculated as `f(n) = g(n) + h(n)`.

#### How it Works:

A* maintains two lists:

*   **Open List**: Contains nodes that have been visited but not yet evaluated.
*   **Closed List**: Contains nodes that have already been evaluated.

The algorithm iteratively selects the node with the lowest `f(n)` value from the open list, moves it to the closed list, and then examines its neighbors. For each neighbor, it calculates `g(n)` and `h(n)` to update `f(n)`, and adds or updates the neighbor in the open list if a shorter path is found.

The search continues until the goal node is reached or the open list becomes empty (meaning no path exists).

```python
import heapq

def a_star(graph, start, goal, heuristic):
    """
    Implements the A* search algorithm.

    Args:
        graph (dict): A dictionary representing the graph where keys are nodes
                      and values are dictionaries of neighbors with their costs.
                      Example: {'A': {'B': 1, 'C': 4}}
        start: The starting node.
        goal: The goal node.
        heuristic (dict): A dictionary mapping each node to its heuristic cost
                          to the goal.
                          Example: {'A': 7, 'B': 6}

    Returns:
        list: The path from start to goal, or None if no path is found.
        int: The total cost of the path, or float('inf') if no path is found.
    """

    # The open list stores (f_cost, node, path) tuples
    open_list = [(0 + heuristic[start], start, [start])]
    heapq.heapify(open_list)

    # g_cost stores the actual cost from start to a node
    g_cost = {node: float('inf') for node in graph}
    g_cost[start] = 0

    # Keep track of visited nodes to avoid redundant processing
    closed_list = set()

    while open_list:
        current_f_cost, current_node, current_path = heapq.heappop(open_list)

        if current_node == goal:
            return current_path, g_cost[goal]

        if current_node in closed_list:
            continue

        closed_list.add(current_node)

        for neighbor, cost in graph[current_node].items():
            # Calculate the g_cost for the neighbor through the current_node
            new_g_cost = g_cost[current_node] + cost

            # If we found a shorter path to the neighbor
            if new_g_cost < g_cost[neighbor]:
                g_cost[neighbor] = new_g_cost
                new_f_cost = new_g_cost + heuristic[neighbor]
                new_path = current_path + [neighbor]
                heapq.heappush(open_list, (new_f_cost, neighbor, new_path))

    return None, float('inf') # No path found

# Example Usage:

# Define the graph
graph = {
    'A': {'B': 1, 'C': 4},
    'B': {'D': 5, 'E': 2},
    'C': {'F': 2},
    'D': {'G': 3},
    'E': {'G': 2},
    'F': {'G': 1},
    'G': {}
}

# Define the heuristic values (estimated cost from node to goal 'G')
heuristic = {
    'A': 7,
    'B': 6,
    'C': 3,
    'D': 2,
    'E': 1,
    'F': 0,
    'G': 0
}

start_node = 'A'
goal_node = 'G'

path, cost = a_star(graph, start_node, goal_node, heuristic)

if path:
    print(f"Shortest path from {start_node} to {goal_node}: {path}")
    print(f"Total cost: {cost}")
else:
    print(f"No path found from {start_node} to {goal_node}")

# Another example with a different graph structure (e.g., a grid-like scenario)
print("\n--- Grid-like Example ---")
grid_graph = {
    'S': {'A': 1, 'B': 4},
    'A': {'C': 2, 'D': 5},
    'B': {'E': 1},
    'C': {'G': 3},
    'D': {'G': 1},
    'E': {'G': 2},
    'G': {}
}

grid_heuristic = {
    'S': 6,
    'A': 4,
    'B': 2,
    'C': 3,
    'D': 1,
    'E': 1,
    'G': 0
}

start_grid = 'S'
goal_grid = 'G'

path_grid, cost_grid = a_star(grid_graph, start_grid, goal_grid, grid_heuristic)

if path_grid:
    print(f"Shortest path from {start_grid} to {goal_grid}: {path_grid}")
    print(f"Total cost: {cost_grid}")
else:
    print(f"No path found from {start_grid} to {goal_grid}")
```
