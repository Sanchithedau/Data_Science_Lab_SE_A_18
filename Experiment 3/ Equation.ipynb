### Solving a system of linear equations using NumPy

We have the following system of linear equations:

$2x + 3y = 10$

$4x + 5y = 20$

This system can be represented in matrix form as $Ax = B$, where:

$A = \begin{pmatrix} 2 & 3 \\ 4 & 5 \end{pmatrix}$, $x = \begin{pmatrix} x \\ y \end{pmatrix}$, and $B = \begin{pmatrix} 10 \\ 20 \end{pmatrix}$

```python
import numpy as np

# Define the coefficient matrix A
A = np.array([[2, 3], [4, 5]])

# Define the constant vector B
B = np.array([10, 20])

# Solve the system of equations Ax = B for x
x = np.linalg.solve(A, B)

print(f"The solution to the system of equations is: x = {x[0]:.2f}, y = {x[1]:.2f}")
```
