```python
import numpy as np

print("Enter the 9 elements for the 3x3 matrix, separated by spaces or newlines:")
matrix_elements = []
for i in range(3):
    row = list(map(float, input(f"Enter row {i+1} elements: ").split()))
    if len(row) != 3:
        raise ValueError("Each row must contain exactly 3 elements.")
    matrix_elements.extend(row)

# Reshape the list into a 3x3 NumPy array
matrix = np.array(matrix_elements).reshape(3, 3)

print("\nYour 3x3 Matrix:")
print(matrix)
```

```python
import numpy as np

# Assuming 'matrix' from the previous cell is available
determinant = np.linalg.det(matrix)

print(f"\nThe determinant of the matrix is: {determinant}")
```
