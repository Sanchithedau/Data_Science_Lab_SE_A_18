```python
import numpy as np

def get_matrix_input():
    """Gets matrix input from the user."""
    while True:
        try:
            rows = int(input("Enter the number of rows for the matrix: "))
            cols = int(input("Enter the number of columns for the matrix: "))
            if rows <= 0 or cols <= 0:
                print("Number of rows and columns must be positive. Please try again.")
                continue
            break
        except ValueError:
            print("Invalid input. Please enter an integer for dimensions.")

    if rows != cols:
        print("\nWarning: Inverse can only be found for square matrices. Please ensure rows == columns.")
        # Optionally, you could exit or re-prompt here if only square matrices are strictly allowed.

    matrix = []
    print(f"\nEnter the elements of the {rows}x{cols} matrix row by row:")
    for i in range(rows):
        while True:
            try:
                row_input = input(f"Enter elements for row {i + 1} (space-separated): ")
                current_row = list(map(float, row_input.split()))
                if len(current_row) != cols:
                    print(f"Expected {cols} elements, but got {len(current_row)}. Please re-enter row {i + 1}.")
                else:
                    matrix.append(current_row)
                    break
            except ValueError:
                print("Invalid input. Please enter numbers separated by spaces.")
    return np.array(matrix)

# Get matrix from user
A = get_matrix_input()

print("\nOriginal Matrix A:")
display(A)

# Check if the matrix is square
if A.shape[0] != A.shape[1]:
    print("\nError: Cannot find the inverse of a non-square matrix.")
else:
    try:
        # Calculate the inverse
        A_inv = np.linalg.inv(A)

        print("\nInverse Matrix A_inv:")
        display(A_inv)

        # Verify (optional): A * A_inv should be close to identity matrix
        print("\nVerification (A * A_inv):")
        display(np.dot(A, A_inv))

    except np.linalg.LinAlgError as e:
        print(f"\nError: The matrix is singular or not invertible. ({e})")
    except Exception as e:
        print(f"\nAn unexpected error occurred: {e}")
```
