A simple expert system can be built using if-else statements (rules) to make decisions based on input conditions. Here's an example that suggests clothing based on weather conditions.

```python
def simple_weather_expert_system(temperature, precipitation):
    """
    A simple expert system to suggest clothing based on weather conditions.

    Args:
        temperature (int): The current temperature in Celsius.
        precipitation (str): Type of precipitation (e.g., 'none', 'rain', 'snow').

    Returns:
        str: A clothing suggestion.
    """
    if precipitation == 'snow':
        return "Wear a heavy coat, hat, and gloves. It's snowing!"
    elif precipitation == 'rain':
        return "Don't forget your raincoat and umbrella. It's raining."
    elif temperature < 10:
        return "It's cold! Wear a warm jacket and layers."
    elif temperature < 20:
        return "It's cool. A light jacket or sweater should be fine."
    else:
        return "It's warm and sunny. T-shirt and shorts are good!"

# --- Test the expert system ---
print("Scenario 1: Cold and no precipitation")
print(simple_weather_expert_system(temperature=5, precipitation='none'))

print("\nScenario 2: Warm and no precipitation")
print(simple_weather_expert_system(temperature=25, precipitation='none'))

print("\nScenario 3: Raining")
print(simple_weather_expert_system(temperature=15, precipitation='rain'))

print("\nScenario 4: Snowing")
print(simple_weather_expert_system(temperature=-2, precipitation='snow'))
```

### How it works:

1.  **Rules:** The system uses `if-elif-else` statements, where each condition acts as a 'rule'.
2.  **Inputs:** It takes `temperature` and `precipitation` as inputs.
3.  **Decision-making:** It evaluates the rules in order. The first rule that matches the input conditions determines the output.
4.  **Output:** It returns a simple string suggesting appropriate clothing.

This is a very basic example. Real-world expert systems can have hundreds or thousands of rules, use more complex logic, and might incorporate certainty factors or fuzzy logic.
