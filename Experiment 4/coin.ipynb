```python
import random

def coin_toss_simulation(num_tosses):
    heads = 0
    tails = 0
    results = []

    for _ in range(num_tosses):
        # 0 for tails, 1 for heads
        if random.randint(0, 1) == 1:
            heads += 1
            results.append('Heads')
        else:
            tails += 1
            results.append('Tails')

    print(f"Simulation of {num_tosses} coin tosses:")
    print(f"Heads: {heads}")
    print(f"Tails: {tails}")

    prob_heads = heads / num_tosses
    prob_tails = tails / num_tosses

    print(f"Probability of Heads: {prob_heads:.2f}")
    print(f"Probability of Tails: {prob_tails:.2f}")

    return results, prob_heads, prob_tails

# Run the simulation with 1000 tosses
num_tosses = 1000
sim_results, prob_h, prob_t = coin_toss_simulation(num_tosses)
```

### Code Explanation:

1.  **`import random`**: This line imports the `random` module, which is used to generate random numbers for the coin toss.
2.  **`coin_toss_simulation(num_tosses)` function**:
    *   Takes `num_tosses` as an argument, which is the number of times the coin will be tossed.
    *   Initializes `heads` and `tails` counters to 0.
    *   `results` is an empty list to store the outcome of each toss.
    *   **Looping through tosses**: The `for` loop runs `num_tosses` times.
    *   **`random.randint(0, 1)`**: This generates either 0 or 1 randomly. We assign 1 to 'Heads' and 0 to 'Tails'.
    *   Based on the random number, the `heads` or `tails` counter is incremented, and the outcome ('Heads' or 'Tails') is added to the `results` list.
    *   **Printing Results**: After all tosses, it prints the total number of heads and tails.
    *   **Calculating Probabilities**: `prob_heads` and `prob_tails` are calculated by dividing the count of heads/tails by the total number of tosses.
    *   The probabilities are printed, formatted to two decimal places.
    *   The function returns the `results` list and the calculated probabilities.
3.  **`num_tosses = 1000`**: Sets the number of coin tosses for the simulation to 1000.
4.  **`sim_results, prob_h, prob_t = coin_toss_simulation(num_tosses)`**: Calls the simulation function and stores its returned values.
